```python
import numpy as np
```

```python


# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Array created:", temperatures_w1)
```

```python
# 2. Inspection and Properties
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# 3. Array Operations
# Convert Celsius to Fahrenheit using (Celsius * 9/5) + 32
temperatures_w1_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_w1_f)

print("Maximum temperature:", np.max(temperatures_w1))
print("Minimum temperature:", np.min(temperatures_w1))
print("Mean temperature:", np.mean(temperatures_w1))
```

```python
# 4. Array Slicing and Indexing
# Extract first three days
print("First three days:", temperatures_w1[:3])

# Extract the weekend (last two days)
print("Weekend:", temperatures_w1[-2:])

# Extract the middle three days (indices 2, 3, 4)
print("Middle three days:", temperatures_w1[2:5])
```

```python


# --- NumPy 2D Array Operations ---

# 5. Create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])
print("2D Array Created:\n", temperatures)
```

```python
# 6. Inspect and Slice the 2D Array
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

# Extract temperatures for each week
print("Week 1 temperatures:", temperatures[0, :])
print("Week 2 temperatures:", temperatures[1, :])

# Extract the weekends (last two days) for both weeks
print("Weekends for both weeks:\n", temperatures[:, -2:])
```

```python
import pandas as pd
# --- Pandas Series Operations ---

# 1. Creating Pandas Series
marks_data = [95, 92, 89, 85, 80]
custom_indices = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
marks = pd.Series(marks_data, index=custom_indices)

print("\nOriginal Pandas Series:\n", marks)
```

```python
# 2. Indexing and Slicing

# Integer index position for the 1st rank student
print("\nMark of 1st rank student:", marks.iloc[0]) 

# loc accessor for the top 3 ranks
print("Marks of top 3 ranks:\n", marks.loc[['Rank1', 'Rank2', 'Rank3']]) 

# iloc accessor for the 3rd rank student
print("Mark of 3rd rank student:", marks.iloc[2]) 

# Boolean mask to filter marks greater than 90
print("Ranks with marks > 90:\n", marks[marks > 90])
```

```python
# 3. Manipulating Series

# Modify the mark of the 1st rank to 100
marks.loc['Rank1'] = 100 

# Remove the entry corresponding to the last rank
marks = marks.drop('Rank5') 

# Compute the CGPA by dividing each mark by 10
cgpa = marks / 10 

print("\nModified Series:\n", marks)
print("\nCGPA Series:\n", cgpa)
```

```python


# 1. Creating Pandas DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110], 
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)

print("--- DataFrame ---")
print(transactions)
```

```python
# 2. Data Exploration
print("\n--- Basic Information ---")
print("Head:\n", transactions.head())
print("\nTail:\n", transactions.tail())
print("\nShape:", transactions.shape)
print("\nColumn Names:", transactions.columns.tolist())
print("\nData Types:\n", transactions.dtypes)
```

```python
print("\n--- Specific Columns ---")
# Display only 'ProductCategory' and 'Amount' columns
print(transactions[['ProductCategory', 'Amount']])

print("\n--- Last 3 Columns using iloc ---")
# Retrieve the last 3 columns using iloc
print(transactions.iloc[:, -3:])
```

```python
print("\n--- Filtered Rows ---")
# Filter rows where Region is 'North' and Amount > 200
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

print("\n--- Value Counts ---")
# Find value counts for 'ProductCategory' column
print(transactions['ProductCategory'].value_counts())

print("\n--- Unique Values ---")
# Find unique values in 'Region' column
print(transactions['Region'].unique())

print("\n--- Group By ---")
# Group by 'Region' and find the mean amount
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. Manipulating the DataFrame
print("\n--- DataFrame Manipulation ---")

# Modify 'Amount' for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

# Add a new column 'Discount' calculating 10% of 'Amount'
transactions['Discount'] = transactions['Amount'] * 0.10

# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Delete the 'Discount' column
transactions = transactions.drop(columns=['Discount'])

print("Final DataFrame after manipulations:\n", transactions)
```
